```python
import builtins

print("--- All Built-in Names ---")
# Use dir(__builtins__) to get a list of all built-in names
all_builtins = dir(builtins)
print(f"Total built-in names: {len(all_builtins)}")
print(all_builtins[:10]) # Displaying first 10 for brevity

print("\n--- Callable Built-in Functions/Methods ---")
# Filter for callable built-ins (functions, methods, classes)
callable_builtins = [name for name in all_builtins if callable(getattr(builtins, name))]
print(f"Total callable built-in functions/methods: {len(callable_builtins)}")

# Display some of them
for i, name in enumerate(callable_builtins):
    print(name)
    if i >= 20: # Displaying first 20 callable built-ins
        break
```

A Python module is essentially a `.py` file containing Python code. You can define functions at the module level, and also define classes. Within these classes, you can define methods. This structure helps organize your code. Below is an example where:

1.  `greet_user` is a module-level function.
2.  `UserManager` is a class defined within the 'module' (this code cell).
3.  `display_greeting` is a method of the `UserManager` class, and it calls the `greet_user` function.

```python
# This entire cell can be thought of as a 'module'

# Function defined directly in the 'module' (global scope of this cell)
def greet_user(name):
    """A simple function to greet a user."""
    return f"Hello, {name}! Welcome to the module."

# Class defined in the 'module'
class UserManager:
    """A class to manage user-related operations."""

    def __init__(self, username):
        self.username = username

    # Method defined inside the class, which is inside the 'module'
    def display_greeting(self):
        """A method that uses the module-level greet_user function."""
        message = greet_user(self.username) # Calling the module-level function
        print(f"Inside UserManager method: {message}")

# --- Demonstration of usage ---

# Use the module-level function
print(greet_user("Alice"))

# Create an instance of the class and use its method
user_manager = UserManager("Bob")
user_manager.display_greeting()

# You can also access the function directly via the module if it were imported
# (e.g., if this code was in a file named `my_module.py` and you did `import my_module`)
# print(my_module.greet_user("Charlie"))
```
